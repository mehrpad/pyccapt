```python
# Activate intractive functionality of matplotlib
%matplotlib notebook
import os
from ipywidgets import widgets
from plotly.offline import init_notebook_mode
from plotly.offline import iplot

from pyccapt.calibration.leap_tools import leap_tools
from pyccapt.calibration.leap_tools import cloud_plotter
from pyccapt.calibration.data_tools import dataset_path_qt
```

```python
button = widgets.Button(
    description='Load Dataset'
)

@button.on_click
def open_file_on_click(b):
    """
    Event handler for the button click event.

    This function is triggered when the 'Load Dataset' button is clicked.
    It assigns the selected dataset path to the global variable 'dataset_path'.

    Args:
        b (ipywidgets.Button): The button object that triggered the event.

    Returns:
        None
    """
    global dataset_path
    dataset_path = dataset_path_qt.gui_fname().decode('ASCII')

button
```

```python
button_range = widgets.Button(
    description='Load Range'
)

try:
    del range_path
except:
    pass

@button_range.on_click
def open_file_on_click(b):
    """
    Event handler for the 'Load Range' button click event.

    This function is triggered when the 'Load Range' button is clicked.
    It assigns the selected range file path to the global variable 'range_path'.

    Args:
        b (ipywidgets.Button): The button object that triggered the event.

    Returns:
        None
    """
    global range_path
    range_path = dataset_path_qt.gui_fname().decode('ASCII')

button_range
```

```python
# Set the absolute path to the desired directory
p = os.path.abspath(os.path.join("", "../../../.."))

# Set the paths for the input data and result output
path = p + '\\tests\\data\\data_leap_test\\'
result_path = p + '\\results\\'

# Create the result directory if it doesn't exist
if not os.path.isdir(result_path):
    os.makedirs(result_path, mode=0o777, exist_ok=True)

# Preprocess positional data for cloud plotting
pos_elements = cloud_plotter.pre_data(path + 'voldata.pos', path + 'rangefile.rrng', input_type='pos')

# Read ions and range values from a file
ions, rrngs = leap_tools.read_rrng(path + 'rangefile.rrng', return_tables=True)

# Extract specific columns and set the index for positional elements
pos_elements_l = pos_elements[['x (nm)', 'y (nm)', 'z (nm)', 'element', 'colour']].set_index("element")

# Print the ions
print(ions)

# Set the phases for the filename
phases = ['H', 'C', 'O', 'Ca', 'Ga', 'Na', 'N']

# Join the phases with underscores to create the filename
filename = '_'.join(phases)
```

```python
# plot the 2d projection of the data
cloud_plotter.cloud_plotter(pos_elements_l, phases, result_path, filename, plot_type='projection')
```

```python
# Initialise the plotly notebook mode
init_notebook_mode(connected=True)
# Create the plotly figure from the data
fig = cloud_plotter.cloud_plotter(pos_elements_l, phases, result_path, filename, plot_type='cloud')
# plotly.offline.plot(fig, filename=result_path + '3d.html', show_link=True,
#                     auto_open=False)
iplot(fig)
```
